```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# Read the data
crop_fulldata = pd.read_csv("/content/Agri_yield_prediction_30k_realistic.csv")

# Round years to nearest integer
crop_fulldata['Year_Rounded'] = crop_fulldata['Year'].round().astype(int)

# Calculate average rainfall per year
yearly_rainfall = crop_fulldata.groupby('Year_Rounded')['Rainfall'].mean().reset_index()
yearly_rainfall.columns = ['Year', 'Rainfall']
yearly_rainfall = yearly_rainfall.sort_values('Year')

print(f"Year range: {yearly_rainfall['Year'].min():.0f} - {yearly_rainfall['Year'].max():.0f}")
print(f"Average Rainfall: {yearly_rainfall['Rainfall'].mean():.2f} mm")

# Create figure with subplots for different visualizations
fig = plt.figure(figsize=(16, 10))

# 1. Main trend plot with moving average
ax1 = plt.subplot(2, 2, 1)
ax1.plot(yearly_rainfall['Year'], yearly_rainfall['Rainfall'],
         marker='o', linewidth=1.5, markersize=5,
         color='#5B9BD5', alpha=0.6, label='Annual Rainfall')

# Add 10-year moving average
if len(yearly_rainfall) >= 10:
    window = 10
    yearly_rainfall['MA'] = yearly_rainfall['Rainfall'].rolling(window=window, center=True).mean()
    ax1.plot(yearly_rainfall['Year'], yearly_rainfall['MA'],
             linewidth=3, color='#E74C3C', label=f'{window}-Year Moving Average')

ax1.set_xlabel('Year', fontsize=11, fontweight='bold')
ax1.set_ylabel('Rainfall (mm)', fontsize=11, fontweight='bold')
ax1.set_title('Annual Rainfall with Moving Average', fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3, linestyle='--')
ax1.legend(loc='best', fontsize=9)
ax1.set_ylim(bottom=0)

# 2. Scatter plot with trend line
ax2 = plt.subplot(2, 2, 2)
ax2.scatter(yearly_rainfall['Year'], yearly_rainfall['Rainfall'],
            alpha=0.5, s=30, color='#3498DB', edgecolors='#2C3E50', linewidth=0.5)

# Add linear regression line
z = np.polyfit(yearly_rainfall['Year'], yearly_rainfall['Rainfall'], 1)
p = np.poly1d(z)
ax2.plot(yearly_rainfall['Year'], p(yearly_rainfall['Year']),
         "r--", linewidth=2.5, label=f'Trend: y={z[0]:.3f}x+{z[1]:.1f}')

ax2.set_xlabel('Year', fontsize=11, fontweight='bold')
ax2.set_ylabel('Rainfall (mm)', fontsize=11, fontweight='bold')
ax2.set_title('Rainfall Trend Analysis', fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3, linestyle='--')
ax2.legend(loc='best', fontsize=9)
ax2.set_ylim(bottom=0)

# 3. Histogram of rainfall distribution
ax3 = plt.subplot(2, 2, 3)
n, bins, patches = ax3.hist(yearly_rainfall['Rainfall'], bins=30,
                             color='#16A085', alpha=0.7, edgecolor='black')
ax3.axvline(yearly_rainfall['Rainfall'].mean(), color='red',
            linestyle='dashed', linewidth=2, label=f'Mean: {yearly_rainfall["Rainfall"].mean():.1f} mm')
ax3.axvline(yearly_rainfall['Rainfall'].median(), color='orange',
            linestyle='dashed', linewidth=2, label=f'Median: {yearly_rainfall["Rainfall"].median():.1f} mm')

ax3.set_xlabel('Rainfall (mm)', fontsize=11, fontweight='bold')
ax3.set_ylabel('Frequency', fontsize=11, fontweight='bold')
ax3.set_title('Rainfall Distribution', fontsize=13, fontweight='bold')
ax3.legend(loc='best', fontsize=9)
ax3.grid(True, alpha=0.3, axis='y')

# 4. Box plot by decade
ax4 = plt.subplot(2, 2, 4)
yearly_rainfall['Decade'] = (yearly_rainfall['Year'] // 10) * 10
decades = sorted(yearly_rainfall['Decade'].unique())

# Only show decades with enough data
decade_data = [yearly_rainfall[yearly_rainfall['Decade'] == d]['Rainfall'].values
               for d in decades if len(yearly_rainfall[yearly_rainfall['Decade'] == d]) >= 3]
decade_labels = [f"{d}s" for d in decades
                 if len(yearly_rainfall[yearly_rainfall['Decade'] == d]) >= 3]

bp = ax4.boxplot(decade_data, labels=decade_labels, patch_artist=True,
                 boxprops=dict(facecolor='#3498DB', alpha=0.7),
                 medianprops=dict(color='red', linewidth=2),
                 whiskerprops=dict(color='#2C3E50'),
                 capprops=dict(color='#2C3E50'))

ax4.set_xlabel('Decade', fontsize=11, fontweight='bold')
ax4.set_ylabel('Rainfall (mm)', fontsize=11, fontweight='bold')
ax4.set_title('Rainfall Distribution by Decade', fontsize=13, fontweight='bold')
ax4.grid(True, alpha=0.3, axis='y')
plt.setp(ax4.xaxis.get_majorticklabels(), rotation=45, ha='right')

plt.suptitle('Comprehensive Rainfall Analysis', fontsize=16, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

# Print detailed statistics
print("\n" + "="*50)
print("DETAILED STATISTICS")
print("="*50)

print(f"\nRainfall Statistics:")
print(f"  Mean:     {yearly_rainfall['Rainfall'].mean():.2f} mm")
print(f"  Median:   {yearly_rainfall['Rainfall'].median():.2f} mm")
print(f"  Std Dev:  {yearly_rainfall['Rainfall'].std():.2f} mm")
print(f"  Min:      {yearly_rainfall['Rainfall'].min():.2f} mm (Year: {yearly_rainfall.loc[yearly_rainfall['Rainfall'].idxmin(), 'Year']:.0f})")
print(f"  Max:      {yearly_rainfall['Rainfall'].max():.2f} mm (Year: {yearly_rainfall.loc[yearly_rainfall['Rainfall'].idxmax(), 'Year']:.0f})")

# Trend analysis
slope, intercept, r_value, p_value, std_err = stats.linregress(yearly_rainfall['Year'], yearly_rainfall['Rainfall'])
print(f"\nTrend Analysis:")
print(f"  Slope:        {slope:.4f} mm/year")
print(f"  R-squared:    {r_value**2:.4f}")
print(f"  P-value:      {p_value:.4e}")
if p_value < 0.05:
    trend_dir = "increasing" if slope > 0 else "decreasing"
    print(f"  → Significant {trend_dir} trend detected!")
else:
    print(f"  → No significant trend detected")

# Decade comparison
print(f"\nAverage Rainfall by Decade:")
decade_avg = yearly_rainfall.groupby('Decade')['Rainfall'].agg(['mean', 'count'])
for decade, row in decade_avg.iterrows():
    if row['count'] >= 3:
        print(f"  {int(decade)}s: {row['mean']:.2f} mm (n={int(row['count'])} years)")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

# Read the data
crop_fulldata = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/graduation project/Agri_yield_prediction_30k_realistic.csv")

print("Dataset shape:", crop_fulldata.shape)
print("\nColumn names:")
print(crop_fulldata.columns.tolist())

# Check if EVI and NDVI columns exist
has_evi = 'EVI' in crop_fulldata.columns
has_ndvi = 'NDVI' in crop_fulldata.columns

if not has_evi or not has_ndvi:
    print("\n⚠️ Searching for EVI/NDVI columns...")
    for col in crop_fulldata.columns:
        if 'evi' in col.lower():
            print(f"Found EVI column: '{col}'")
            has_evi = True
            crop_fulldata['EVI'] = crop_fulldata[col]
        if 'ndvi' in col.lower():
            print(f"Found NDVI column: '{col}'")
            has_ndvi = True
            crop_fulldata['NDVI'] = crop_fulldata[col]

if has_evi and has_ndvi:
    # Original data statistics
    print("\n" + "="*60)
    print("ORIGINAL DATA")
    print("="*60)
    print(f"Total records: {len(crop_fulldata)}")
    print(f"EVI range: {crop_fulldata['EVI'].min():.4f} to {crop_fulldata['EVI'].max():.4f}")
    print(f"NDVI range: {crop_fulldata['NDVI'].min():.4f} to {crop_fulldata['NDVI'].max():.4f}")

    # ===== NOISE REMOVAL METHODS =====

    # Method 1: Remove outliers using IQR (Interquartile Range)
    def remove_outliers_iqr(df, column):
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

    # Method 2: Remove values outside valid ranges
    # NDVI valid range: -1 to 1 (typically 0 to 1 for vegetation)
    # EVI valid range: -1 to 1 (typically 0 to 1 for vegetation)
    crop_cleaned = crop_fulldata.copy()

    # Remove invalid NDVI and EVI values
    crop_cleaned = crop_cleaned[(crop_cleaned['NDVI'] >= -1) & (crop_cleaned['NDVI'] <= 1)]
    crop_cleaned = crop_cleaned[(crop_cleaned['EVI'] >= -1) & (crop_cleaned['EVI'] <= 1)]

    # Remove outliers for EVI
    crop_cleaned = remove_outliers_iqr(crop_cleaned, 'EVI')

    # Remove outliers for NDVI
    crop_cleaned = remove_outliers_iqr(crop_cleaned, 'NDVI')

    # Remove NaN values
    crop_cleaned = crop_cleaned.dropna(subset=['EVI', 'NDVI'])

    # Method 3: Remove data points with extreme differences (inconsistent readings)
    # Calculate residuals from linear regression
    z_evi_ndvi = np.polyfit(crop_cleaned['NDVI'], crop_cleaned['EVI'], 1)
    p_evi_ndvi = np.poly1d(z_evi_ndvi)
    crop_cleaned['predicted_EVI'] = p_evi_ndvi(crop_cleaned['NDVI'])
    crop_cleaned['residual'] = crop_cleaned['EVI'] - crop_cleaned['predicted_EVI']

    # Remove points with high residuals (noise)
    residual_threshold = 2 * crop_cleaned['residual'].std()
    crop_cleaned = crop_cleaned[abs(crop_cleaned['residual']) <= residual_threshold]

    print("\n" + "="*60)
    print("CLEANED DATA (After Noise Removal)")
    print("="*60)
    print(f"Total records: {len(crop_cleaned)}")
    print(f"Removed records: {len(crop_fulldata) - len(crop_cleaned)} ({((len(crop_fulldata) - len(crop_cleaned))/len(crop_fulldata)*100):.1f}%)")
    print(f"EVI range: {crop_cleaned['EVI'].min():.4f} to {crop_cleaned['EVI'].max():.4f}")
    print(f"NDVI range: {crop_cleaned['NDVI'].min():.4f} to {crop_cleaned['NDVI'].max():.4f}")

    # Calculate correlations
    corr_original = crop_fulldata[['NDVI', 'EVI']].corr().iloc[0, 1]
    corr_cleaned = crop_cleaned[['NDVI', 'EVI']].corr().iloc[0, 1]

    print(f"\nCorrelation (Original): {corr_original:.4f}")
    print(f"Correlation (Cleaned):  {corr_cleaned:.4f}")
    print(f"Improvement: {((corr_cleaned - corr_original)/abs(corr_original)*100):.1f}%")

    # ===== VISUALIZATION =====
    fig = plt.figure(figsize=(16, 10))

    # 1. Before cleaning - Scatter plot
    ax1 = plt.subplot(2, 3, 1)
    ax1.scatter(crop_fulldata['NDVI'], crop_fulldata['EVI'],
               alpha=0.3, s=15, c='gray', edgecolors='none', label='All data')

    z_orig = np.polyfit(crop_fulldata['NDVI'], crop_fulldata['EVI'], 1)
    p_orig = np.poly1d(z_orig)
    ndvi_range = np.linspace(crop_fulldata['NDVI'].min(), crop_fulldata['NDVI'].max(), 100)
    ax1.plot(ndvi_range, p_orig(ndvi_range), "r--", linewidth=2.5,
            label=f'Trend: y={z_orig[0]:.2f}x+{z_orig[1]:.2f}')

    ax1.set_xlabel('NDVI', fontsize=11, fontweight='bold')
    ax1.set_ylabel('EVI', fontsize=11, fontweight='bold')
    ax1.set_title(f'Before Cleaning (Corr: {corr_original:.3f})',
                 fontsize=12, fontweight='bold', color='red')
    ax1.legend(fontsize=9)
    ax1.grid(True, alpha=0.3)

    # 2. After cleaning - Scatter plot
    ax2 = plt.subplot(2, 3, 2)
    scatter = ax2.scatter(crop_cleaned['NDVI'], crop_cleaned['EVI'],
                         alpha=0.5, s=20, c=crop_cleaned['EVI'],
                         cmap='YlGn', edgecolors='none')

    z_clean = np.polyfit(crop_cleaned['NDVI'], crop_cleaned['EVI'], 1)
    p_clean = np.poly1d(z_clean)
    ndvi_range_clean = np.linspace(crop_cleaned['NDVI'].min(),
                                   crop_cleaned['NDVI'].max(), 100)
    ax2.plot(ndvi_range_clean, p_clean(ndvi_range_clean), "r--", linewidth=2.5,
            label=f'Trend: y={z_clean[0]:.2f}x+{z_clean[1]:.2f}')

    ax2.set_xlabel('NDVI', fontsize=11, fontweight='bold')
    ax2.set_ylabel('EVI', fontsize=11, fontweight='bold')
    ax2.set_title(f'After Cleaning (Corr: {corr_cleaned:.3f})',
                 fontsize=12, fontweight='bold', color='green')
    ax2.legend(fontsize=9)
    ax2.grid(True, alpha=0.3)
    plt.colorbar(scatter, ax=ax2, label='EVI Value')

    # 3. Density plot of cleaned data
    ax3 = plt.subplot(2, 3, 3)
    from scipy.stats import gaussian_kde

    # Calculate density
    xy = np.vstack([crop_cleaned['NDVI'], crop_cleaned['EVI']])
    z_density = gaussian_kde(xy)(xy)

    scatter2 = ax3.scatter(crop_cleaned['NDVI'], crop_cleaned['EVI'],
                          c=z_density, s=20, cmap='plasma', alpha=0.6)
    ax3.plot(ndvi_range_clean, p_clean(ndvi_range_clean), "r--", linewidth=2.5)

    ax3.set_xlabel('NDVI', fontsize=11, fontweight='bold')
    ax3.set_ylabel('EVI', fontsize=11, fontweight='bold')
    ax3.set_title('Density Plot (Cleaned Data)', fontsize=12, fontweight='bold')
    ax3.grid(True, alpha=0.3)
    plt.colorbar(scatter2, ax=ax3, label='Density')

    # 4. Time series comparison (if Year exists)
    if 'Year' in crop_cleaned.columns:
        ax4 = plt.subplot(2, 3, 4)
        crop_cleaned['Year_Rounded'] = crop_cleaned['Year'].round().astype(int)
        yearly_veg = crop_cleaned.groupby('Year_Rounded')[['EVI', 'NDVI']].mean().reset_index()

        ax4.plot(yearly_veg['Year_Rounded'], yearly_veg['EVI'],
                marker='o', linewidth=2.5, markersize=6, color='#27AE60',
                label='EVI', alpha=0.9)
        ax4.plot(yearly_veg['Year_Rounded'], yearly_veg['NDVI'],
                marker='s', linewidth=2.5, markersize=6, color='#E67E22',
                label='NDVI', alpha=0.9)

        ax4.set_xlabel('Year', fontsize=11, fontweight='bold')
        ax4.set_ylabel('Index Value', fontsize=11, fontweight='bold')
        ax4.set_title('EVI & NDVI Trends (Cleaned)', fontsize=12, fontweight='bold')
        ax4.legend(loc='best', fontsize=10)
        ax4.grid(True, alpha=0.3, linestyle='--')

    # 5. Distribution comparison
    ax5 = plt.subplot(2, 3, 5)
    ax5.hist(crop_fulldata['EVI'], bins=40, alpha=0.5, color='red',
            label='EVI Before', edgecolor='black')
    ax5.hist(crop_cleaned['EVI'], bins=40, alpha=0.7, color='green',
            label='EVI After', edgecolor='black')
    ax5.axvline(crop_fulldata['EVI'].mean(), color='red',
               linestyle='dashed', linewidth=2, alpha=0.5)
    ax5.axvline(crop_cleaned['EVI'].mean(), color='green',
               linestyle='dashed', linewidth=2)
    ax5.set_xlabel('EVI Value', fontsize=11, fontweight='bold')
    ax5.set_ylabel('Frequency', fontsize=11, fontweight='bold')
    ax5.set_title('EVI Distribution Comparison', fontsize=12, fontweight='bold')
    ax5.legend(fontsize=9)
    ax5.grid(True, alpha=0.3, axis='y')

    # 6. Residuals plot
    ax6 = plt.subplot(2, 3, 6)
    ax6.scatter(crop_cleaned['NDVI'], crop_cleaned['residual'],
               alpha=0.5, s=15, color='purple')
    ax6.axhline(y=0, color='r', linestyle='-', linewidth=2)
    ax6.axhline(y=residual_threshold, color='orange', linestyle='--', linewidth=1.5,
               label=f'Threshold: ±{residual_threshold:.3f}')
    ax6.axhline(y=-residual_threshold, color='orange', linestyle='--', linewidth=1.5)

    ax6.set_xlabel('NDVI', fontsize=11, fontweight='bold')
    ax6.set_ylabel('Residuals', fontsize=11, fontweight='bold')
    ax6.set_title('Residuals Distribution', fontsize=12, fontweight='bold')
    ax6.legend(fontsize=9)
    ax6.grid(True, alpha=0.3)

    plt.suptitle('EVI & NDVI Analysis - Noise Removal & Cleaning',
                fontsize=16, fontweight='bold', y=0.995)
    plt.tight_layout()
    plt.show()

    # Print detailed statistics
    print("\n" + "="*60)
    print("DETAILED COMPARISON")
    print("="*60)

    print("\n📊 EVI Statistics:")
    print(f"{'':15s} {'Original':>12s} {'Cleaned':>12s} {'Change':>12s}")
    print(f"{'Mean':15s} {crop_fulldata['EVI'].mean():>12.4f} {crop_cleaned['EVI'].mean():>12.4f} {(crop_cleaned['EVI'].mean()-crop_fulldata['EVI'].mean()):>12.4f}")
    print(f"{'Std Dev':15s} {crop_fulldata['EVI'].std():>12.4f} {crop_cleaned['EVI'].std():>12.4f} {(crop_cleaned['EVI'].std()-crop_fulldata['EVI'].std()):>12.4f}")
    print(f"{'Min':15s} {crop_fulldata['EVI'].min():>12.4f} {crop_cleaned['EVI'].min():>12.4f} {(crop_cleaned['EVI'].min()-crop_fulldata['EVI'].min()):>12.4f}")
    print(f"{'Max':15s} {crop_fulldata['EVI'].max():>12.4f} {crop_cleaned['EVI'].max():>12.4f} {(crop_cleaned['EVI'].max()-crop_fulldata['EVI'].max()):>12.4f}")

    print("\n🌱 NDVI Statistics:")
    print(f"{'':15s} {'Original':>12s} {'Cleaned':>12s} {'Change':>12s}")
    print(f"{'Mean':15s} {crop_fulldata['NDVI'].mean():>12.4f} {crop_cleaned['NDVI'].mean():>12.4f} {(crop_cleaned['NDVI'].mean()-crop_fulldata['NDVI'].mean()):>12.4f}")
    print(f"{'Std Dev':15s} {crop_fulldata['NDVI'].std():>12.4f} {crop_cleaned['NDVI'].std():>12.4f} {(crop_cleaned['NDVI'].std()-crop_fulldata['NDVI'].std()):>12.4f}")
    print(f"{'Min':15s} {crop_fulldata['NDVI'].min():>12.4f} {crop_cleaned['NDVI'].min():>12.4f} {(crop_cleaned['NDVI'].min()-crop_fulldata['NDVI'].min()):>12.4f}")
    print(f"{'Max':15s} {crop_fulldata['NDVI'].max():>12.4f} {crop_cleaned['NDVI'].max():>12.4f} {(crop_cleaned['NDVI'].max()-crop_fulldata['NDVI'].max()):>12.4f}")

    print("\n✅ Noise Removal Summary:")
    print(f"  Records kept: {len(crop_cleaned)}/{len(crop_fulldata)} ({len(crop_cleaned)/len(crop_fulldata)*100:.1f}%)")
    print(f"  Correlation improvement: {corr_original:.4f} → {corr_cleaned:.4f}")
    print(f"  Data quality: {'Excellent' if corr_cleaned > 0.8 else 'Good' if corr_cleaned > 0.6 else 'Moderate'}")

else:
    print("\n❌ Error: Could not find EVI and/or NDVI columns!")
    print("Available columns:", crop_fulldata.columns.tolist())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")

crop_fulldata = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/graduation project/Agri_yield_prediction_30k_realistic.csv")

print("=" * 70)
print("  TEMPERATURE ANALYSIS - Agricultural Intelligence System")
print("=" * 70)
print("\n Basic Temperature Statistics:")
print("-" * 70)
print(f"Mean Temperature:        {crop_fulldata['Temperature'].mean():.2f}°C")
print(f"Median Temperature:      {crop_fulldata['Temperature'].median():.2f}°C")
print(f"Std Deviation:           {crop_fulldata['Temperature'].std():.2f}°C")
print(f"Min Temperature:         {crop_fulldata['Temperature'].min():.2f}°C")
print(f"Max Temperature:         {crop_fulldata['Temperature'].max():.2f}°C")
print(f"25th Percentile (Q1):    {crop_fulldata['Temperature'].quantile(0.25):.2f}°C")
print(f"75th Percentile (Q3):    {crop_fulldata['Temperature'].quantile(0.75):.2f}°C")
print(f"IQR (Q3-Q1):             {crop_fulldata['Temperature'].quantile(0.75) - crop_fulldata['Temperature'].quantile(0.25):.2f}°C")
fig = plt.figure(figsize=(16, 12))
gs = fig.add_gridspec(3, 2, hspace=0.3, wspace=0.3)
ax1 = fig.add_subplot(gs[0, 0])
n, bins, patches = ax1.hist(crop_fulldata['Temperature'], bins=50,
                            color='#FF6B6B', alpha=0.7, edgecolor='black', linewidth=1.2)
cm = plt.cm.RdYlBu_r
norm = plt.Normalize(vmin=crop_fulldata['Temperature'].min(),
                     vmax=crop_fulldata['Temperature'].max())
for i, p in enumerate(patches):
    plt.setp(p, 'facecolor', cm(norm((bins[i] + bins[i+1])/2)))

ax1_twin = ax1.twinx()
crop_fulldata['Temperature'].plot(kind='kde', ax=ax1_twin, color='darkred',
                                  linewidth=3, label='KDE')
ax1.set_xlabel('Temperature (°C)', fontsize=12, fontweight='bold')
ax1.set_ylabel('Frequency', fontsize=12, fontweight='bold')
ax1_twin.set_ylabel('Density', fontsize=12, fontweight='bold')
ax1.set_title('📊 Temperature Distribution with KDE', fontsize=14, fontweight='bold', pad=15)
ax1.grid(alpha=0.3)
ax1_twin.legend(loc='upper right')
ax2 = fig.add_subplot(gs[0, 1])
box = ax2.boxplot(crop_fulldata['Temperature'], vert=True, patch_artist=True,
                  boxprops=dict(facecolor='#4ECDC4', alpha=0.7, linewidth=2),
                  whiskerprops=dict(linewidth=2, color='#2C3E50'),
                  capprops=dict(linewidth=2, color='#2C3E50'),
                  medianprops=dict(linewidth=3, color='#E74C3C'),
                  flierprops=dict(marker='o', markerfacecolor='red', markersize=8, alpha=0.5))
stats_text = f"Mean: {crop_fulldata['Temperature'].mean():.2f}°C\n"
stats_text += f"Median: {crop_fulldata['Temperature'].median():.2f}°C\n"
stats_text += f"Std: {crop_fulldata['Temperature'].std():.2f}°C"
ax2.text(1.3, crop_fulldata['Temperature'].mean(), stats_text,
         fontsize=10, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

ax2.set_ylabel('Temperature (°C)', fontsize=12, fontweight='bold')
ax2.set_title(' Temperature Box Plot & Outliers', fontsize=14, fontweight='bold', pad=15)
ax2.grid(axis='y', alpha=0.3)
ax2.set_xticklabels(['Temperature'])

# 1.3 Violin Plot
ax3 = fig.add_subplot(gs[1, 0])
parts = ax3.violinplot([crop_fulldata['Temperature']], positions=[1],
                       showmeans=True, showmedians=True, widths=0.7)

for pc in parts['bodies']:
    pc.set_facecolor('#45B7D1')
    pc.set_alpha(0.7)
    pc.set_edgecolor('black')
    pc.set_linewidth(2)

parts['cmeans'].set_color('red')
parts['cmeans'].set_linewidth(3)
parts['cmedians'].set_color('darkgreen')
parts['cmedians'].set_linewidth(3)

ax3.set_ylabel('Temperature (°C)', fontsize=12, fontweight='bold')
ax3.set_title(' Temperature Violin Plot', fontsize=14, fontweight='bold', pad=15)
ax3.set_xticks([1])
ax3.set_xticklabels(['Temperature'])
ax3.grid(axis='y', alpha=0.3)
ax4 = fig.add_subplot(gs[1, 1])
stats.probplot(crop_fulldata['Temperature'], dist="norm", plot=ax4)
ax4.get_lines()[0].set_color('#FF6B6B')
ax4.get_lines()[0].set_markersize(5)
ax4.get_lines()[0].set_alpha(0.6)
ax4.get_lines()[1].set_color('blue')
ax4.get_lines()[1].set_linewidth(2)
ax4.set_title(' Q-Q Plot (Normality Check)', fontsize=14, fontweight='bold', pad=15)
ax4.set_xlabel('Theoretical Quantiles', fontsize=11, fontweight='bold')
ax4.set_ylabel('Sample Quantiles', fontsize=11, fontweight='bold')
ax4.grid(alpha=0.3)
ax5 = fig.add_subplot(gs[2, :])
ax5.plot(crop_fulldata.index, crop_fulldata['Temperature'],
         color='#FF6B6B', linewidth=1, alpha=0.6, label='Temperature')
window = 500
rolling_mean = crop_fulldata['Temperature'].rolling(window=window).mean()
ax5.plot(crop_fulldata.index, rolling_mean,
         color='darkred', linewidth=3, label=f'Moving Average ({window} samples)')
mean_temp = crop_fulldata['Temperature'].mean()
std_temp = crop_fulldata['Temperature'].std()

ax5.axhline(y=mean_temp, color='green', linestyle='--', linewidth=2,
            label=f'Mean: {mean_temp:.2f}°C')
ax5.axhline(y=mean_temp + std_temp, color='orange', linestyle=':', linewidth=2,
            label=f'Mean + 1σ: {mean_temp + std_temp:.2f}°C')
ax5.axhline(y=mean_temp - std_temp, color='orange', linestyle=':', linewidth=2,
            label=f'Mean - 1σ: {mean_temp - std_temp:.2f}°C')
ax5.fill_between(crop_fulldata.index, mean_temp - std_temp, mean_temp + std_temp,
                 alpha=0.2, color='orange')

ax5.set_xlabel('Sample Index', fontsize=12, fontweight='bold')
ax5.set_ylabel('Temperature (°C)', fontsize=12, fontweight='bold')
ax5.set_title(' Temperature Time Series with Moving Average & Statistical Boundaries',
              fontsize=14, fontweight='bold', pad=15)
ax5.legend(loc='best', fontsize=10)
ax5.grid(alpha=0.3)

plt.suptitle('🌡️ Comprehensive Temperature Analysis Dashboard',
             fontsize=18, fontweight='bold', y=0.995)
plt.show()
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

crop_types = crop_fulldata['Crop_Type'].unique()
temp_by_crop = [crop_fulldata[crop_fulldata['Crop_Type'] == crop]['Temperature'].values
                for crop in crop_types]

bp = ax1.boxplot(temp_by_crop, labels=crop_types, patch_artist=True)

colors = plt.cm.Set3(np.linspace(0, 1, len(crop_types)))
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

ax1.set_xlabel('Crop Type', fontsize=12, fontweight='bold')
ax1.set_ylabel('Temperature (°C)', fontsize=12, fontweight='bold')
ax1.set_title(' Temperature Distribution by Crop Type', fontsize=14, fontweight='bold', pad=15)
ax1.grid(axis='y', alpha=0.3)
ax1.tick_params(axis='x', rotation=45)
for i, crop in enumerate(crop_types):
    crop_data = crop_fulldata[crop_fulldata['Crop_Type'] == crop]['Temperature']
    parts = ax2.violinplot([crop_data], positions=[i], showmeans=True,
                          showmedians=True, widths=0.7)
    for pc in parts['bodies']:
        pc.set_facecolor(colors[i])
        pc.set_alpha(0.7)

ax2.set_xticks(range(len(crop_types)))
ax2.set_xticklabels(crop_types, rotation=45, ha='right')
ax2.set_xlabel('Crop Type', fontsize=12, fontweight='bold')
ax2.set_ylabel('Temperature (°C)', fontsize=12, fontweight='bold')
ax2.set_title(' Temperature Violin Plot by Crop Type', fontsize=14, fontweight='bold', pad=15)
ax2.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
seasons = crop_fulldata['Season'].unique()
temp_by_season = [crop_fulldata[crop_fulldata['Season'] == season]['Temperature'].values
                  for season in seasons]

bp = axes[0, 0].boxplot(temp_by_season, labels=seasons, patch_artist=True)
season_colors = {'Spring': '#90EE90', 'Summer': '#FFD700', 'Fall': '#FF8C00', 'Winter': '#87CEEB'}
for patch, season in zip(bp['boxes'], seasons):
    patch.set_facecolor(season_colors.get(season, '#CCCCCC'))
    patch.set_alpha(0.7)

axes[0, 0].set_xlabel('Season', fontsize=12, fontweight='bold')
axes[0, 0].set_ylabel('Temperature (°C)', fontsize=12, fontweight='bold')
axes[0, 0].set_title(' Temperature by Season', fontsize=14, fontweight='bold', pad=15)
axes[0, 0].grid(axis='y', alpha=0.3)
season_avg = crop_fulldata.groupby('Season')['Temperature'].mean().sort_values()
bars = axes[0, 1].bar(season_avg.index, season_avg.values,
                      color=[season_colors.get(s, '#CCCCCC') for s in season_avg.index],
                      alpha=0.7, edgecolor='black', linewidth=2)

for bar in bars:
    height = bar.get_height()
    axes[0, 1].text(bar.get_x() + bar.get_width()/2., height,
                    f'{height:.1f}°C',
                    ha='center', va='bottom', fontsize=11, fontweight='bold')

axes[0, 1].set_xlabel('Season', fontsize=12, fontweight='bold')
axes[0, 1].set_ylabel('Average Temperature (°C)', fontsize=12, fontweight='bold')
axes[0, 1].set_title(' Average Temperature by Season', fontsize=14, fontweight='bold', pad=15)
axes[0, 1].grid(axis='y', alpha=0.3)
for season in seasons:
    season_data = crop_fulldata[crop_fulldata['Season'] == season]['Temperature']
    season_data.plot(kind='kde', ax=axes[1, 0], linewidth=2.5,
                    label=season, alpha=0.7)

axes[1, 0].set_xlabel('Temperature (°C)', fontsize=12, fontweight='bold')
axes[1, 0].set_ylabel('Density', fontsize=12, fontweight='bold')
axes[1, 0].set_title(' Temperature Density by Season', fontsize=14, fontweight='bold', pad=15)
axes[1, 0].legend(fontsize=11)
axes[1, 0].grid(alpha=0.3)
season_stats = crop_fulldata.groupby('Season')['Temperature'].agg(['mean', 'std', 'min', 'max'])
sns.heatmap(season_stats.T, annot=True, fmt='.2f', cmap='RdYlBu_r',
            linewidths=2, cbar_kws={'label': 'Temperature (°C)'}, ax=axes[1, 1])
axes[1, 1].set_title(' Temperature Statistics Heatmap by Season',
                     fontsize=14, fontweight='bold', pad=15)
axes[1, 1].set_xlabel('Season', fontsize=12, fontweight='bold')
axes[1, 1].set_ylabel('Statistic', fontsize=12, fontweight='bold')

plt.suptitle('🌡️ Temperature Analysis by Season', fontsize=16, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))


scatter = ax1.scatter(crop_fulldata['Temperature'], crop_fulldata['Yield'],
                     c=crop_fulldata['Temperature'], cmap='coolwarm',
                     alpha=0.5, s=30, edgecolors='black', linewidth=0.5)

z = np.polyfit(crop_fulldata['Temperature'], crop_fulldata['Yield'], 1)
p = np.poly1d(z)
ax1.plot(crop_fulldata['Temperature'], p(crop_fulldata['Temperature']),
         "r--", linewidth=3, label=f'Trend Line: y={z[0]:.2f}x+{z[1]:.2f}')
correlation = crop_fulldata['Temperature'].corr(crop_fulldata['Yield'])
ax1.text(0.05, 0.95, f'Correlation: {correlation:.3f}',
         transform=ax1.transAxes, fontsize=12, fontweight='bold',
         verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

ax1.set_xlabel('Temperature (°C)', fontsize=12, fontweight='bold')
ax1.set_ylabel('Yield', fontsize=12, fontweight='bold')
ax1.set_title('🌡️ Temperature vs Yield Relationship', fontsize=14, fontweight='bold', pad=15)
ax1.legend()
ax1.grid(alpha=0.3)
plt.colorbar(scatter, ax=ax1, label='Temperature (°C)')
hexbin = ax2.hexbin(crop_fulldata['Temperature'], crop_fulldata['Yield'],
                    gridsize=30, cmap='YlOrRd', mincnt=1)
ax2.set_xlabel('Temperature (°C)', fontsize=12, fontweight='bold')
ax2.set_ylabel('Yield', fontsize=12, fontweight='bold')
ax2.set_title('🔥 Temperature vs Yield Density Map', fontsize=14, fontweight='bold', pad=15)
plt.colorbar(hexbin, ax=ax2, label='Count')
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()
if 'Region' in crop_fulldata.columns:
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

    regions = crop_fulldata['Region'].unique()
    region_avg = crop_fulldata.groupby('Region')['Temperature'].mean().sort_values(ascending=False)

    # Bar plot
    bars = ax1.barh(range(len(region_avg)), region_avg.values,
                    color=plt.cm.plasma(np.linspace(0, 1, len(region_avg))),
                    alpha=0.7, edgecolor='black', linewidth=1.5)

    ax1.set_yticks(range(len(region_avg)))
    ax1.set_yticklabels(region_avg.index, fontsize=10)
    ax1.set_xlabel('Average Temperature (°C)', fontsize=12, fontweight='bold')
    ax1.set_title('Average Temperature by Region', fontsize=14, fontweight='bold', pad=15)
    ax1.grid(axis='x', alpha=0.3)

    for i, (bar, val) in enumerate(zip(bars, region_avg.values)):
        ax1.text(val, i, f' {val:.1f}°C',
                va='center', fontsize=10, fontweight='bold')

    # Box plot by Region
    temp_by_region = [crop_fulldata[crop_fulldata['Region'] == region]['Temperature'].values
                      for region in regions]
    bp = ax2.boxplot(temp_by_region, labels=regions, patch_artist=True)

    colors = plt.cm.viridis(np.linspace(0, 1, len(regions)))
    for patch, color in zip(bp['boxes'], colors):
        patch.set_facecolor(color)
        patch.set_alpha(0.7)

    ax2.set_xlabel('Region', fontsize=12, fontweight='bold')
    ax2.set_ylabel('Temperature (°C)', fontsize=12, fontweight='bold')
    ax2.set_title(' Temperature Distribution by Region', fontsize=14, fontweight='bold', pad=15)
    ax2.tick_params(axis='x', rotation=45)
    ax2.grid(axis='y', alpha=0.3)

    plt.tight_layout()
    plt.show()

print("\n" + "=" * 70)
print(" Temperature Analysis Complete!")
print("=" * 70)
```
